#**Section B**

**Section B – Filtering & Conditional Selection**

1. Select all students who completed the test preparation course.

2. Find all students who scored above 70 in math.

3. Retrieve female students who scored at least 65 in all three subjects (use the & operator).

4. Count how many students have free/reduced lunch.

```python
#1 Students who completed the test preparation course.

students_completed = students[students["test preparation course"] == "completed"]
print(students_completed)
f'{len(students_completed)} students completed the test preparation course.'
```

```python
students["math score"] < 10
```

```python
#2 Students who scored above 70 in math.

students_below_70 = students[students["math score"] > 70]
print(students_below_70)
f'{len(students_below_70)} students scored above 70 in math.'
```

```python
#3 female students who scored at least 65 in all three subjects (use the & operator).

students_female = students[students["gender"] == "female"]
students_female_above_65 = students_female[(students_female["math score"] > 65) & (students_female["reading score"] > 65) & (students_female["writing score"] > 65)]
print(students_female_above_65)
f'{len(students_female_above_65)} female students scored above 65 in all three subjects.'
```

```python
#4 Count of how many students have free/reduced lunch.

students_lunch = students[students["lunch"] == "free/reduced"]
print(students_lunch)
f'{len(students_lunch)} students have free/reduced lunch.'
```

#**Section C**

**Section C – GroupBy & Aggregation**

1. Compute average math score by gender.

2. Compute average scores (math, reading, writing) by parental level of education.

3. Compare average scores of students with vs without test preparation (note: 'none' refers to students who did not complete the test preparation course).

```python
#1 Average math score by gender

students.groupby("gender")["math score"].mean()
```

```python
#2 Average scores (math, reading, writing) by parental level of education

students.groupby("parental level of education")[["math score", "reading score", "writing score"]].mean()
```

```python
#3 average scores of students with vs without test preparation

students.groupby("test preparation course")[["math score", "reading score", "writing score"]].mean()
```

```python
#1a Histogram Plot of Maths Score

plt.hist(students["math score"])
plt.title("Plot of Math Score")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.show()
```

```python
#1b Histogram Plot of Reading Score

plt.hist(students["reading score"])
plt.title("Plot of Reading Score")
plt.xlabel("Reading Score")
plt.ylabel("Frequency")
plt.show()
```

```python
#1c Histogram Plot of Writing Score

plt.hist(students["writing score"])
plt.title("Plot of Writing Score")
plt.xlabel("Writing Score")
plt.ylabel("Frequency")
plt.show()
```

```python
# bar plot showing average score by gender

avg_score = students.groupby("gender")[["math score", "reading score", "writing score"]].mean().plot(kind="bar")
plt.title("Average Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.show()
```
